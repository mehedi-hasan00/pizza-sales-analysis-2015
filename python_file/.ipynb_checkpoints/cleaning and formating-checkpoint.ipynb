```python
# Fixed Datatype

df['order_date'] = pd.to_datetime(df['order_date'])
df['order_time'] = pd.to_datetime(df['order_time'], format = '%H:%M:%S').dt.time
```

```python
# Time based columns

def get_time_of_day(t):
    hour = t.hour
    if 9<= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 14:
        return "Lunch"
    elif 15 <= hour <= 17:
        return "Afternoon"
    elif 18 <= hour <= 20:
        return "Evening"
    else:
        return "Night"
```

```python
df['time_of_day'] = df['order_time'].apply(get_time_of_day)
```

```python
df['time_of_day']
```

```python
# Day of week

df['day_of_week'] = df['order_date'].dt.day_name()
```

```python
# Weekend/weekday

df['is_weekend'] = df['order_date'].dt.day_of_week.apply(lambda x: 'Weekend' if x>=5 else 'Weekday')
```

```python
df.sample(5)
```

```python
# month name
df['month_name'] = df['order_date'].dt.strftime('%b')
```

```python
# Quarter

df['quarter'] = df['order_date'].dt.quarter.apply(lambda x: f'Q{x}')
```

```python
# Revenue based columns
```

```python
df['unit_price'].unique()
```

```python
# revenue category (based on unit_price)

def get_revenue_category(price):
    if price > 20:
        return "Premium"
    elif price >= 12:
        return "Standard"
    else:
        return "Budget"
```

```python
df['revenue_category'] = df['unit_price'].apply(get_revenue_category)
```

```python
df.head()
```

```python
# Order size

def get_order_size(qty):
    if qty == 1:
        return "Small"
    elif qty == 2:
        return "Medium"
    else:
        return "Large"
```

```python
df['order_size'] = df['quantity'].apply(get_order_size)
```

```python
# Pizza based column
# vegetarian check (ingredients based)

non_veg_keywords = ['chicken','beef','pork','pepperoni','sausage','bacon','ham','salami','anchovies','prosiutto']
def is_vegetarian(ingredient):
    ingredients_lower = str(ingredient).lower()
    for keywords in non_veg_keywords:
        if keywords in ingredients_lower:
            return "Non-Veg"
    return "Veg"
```

```python
df['is_vegetarian'] = df['pizza_ingredients'].apply(is_vegetarian)
```

```python
df.head()
```

```python
# Formatting

# Reorder columns nicely
column_order = [
    "pizza_id", "order_id", "order_date", "order_time",
    "day_of_week", "is_weekend", "month_name", "quarter", "time_of_day",
    "pizza_name_id", "pizza_name", "pizza_category", "pizza_size",
    "pizza_ingredients", "is_vegetarian",
    "quantity", "order_size",
    "unit_price", "total_price", "revenue_category"
]

df = df[column_order]
```

```python
df.head()
```

```python
# capitalize text column
text_cols = ['pizza_name','pizza_category','pizza_size','pizza_ingredients']
for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
# format date nicely

df['order_date'] = df['order_date'].dt.strftime('%Y-%m-%d')
```

```python
df.head()
```

```python
#save output

df.to_csv('pizza_sales_cleaned.csv', index = False)
```
